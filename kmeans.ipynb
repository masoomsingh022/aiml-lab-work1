```python
import numpy as np#Used to create dataset (array format)
from sklearn.cluster import KMeans#Imports K-Means algorithm
import matplotlib.pyplot as plt#Used to draw graphs (elbow + cluster plot)
from sklearn.preprocessing import StandardScaler#Used for scaling data (important for distance)
from sklearn.metrics import silhouette_score#Used to check clustering quality
```

```python
X=np.array([10,20,30,40,50,60]).reshape(-1,1)#must be 2d
#Step 2: Data Cleaning (example - not needed here but shown)
# (no missing values here, so skipped)
#remove null values
#remove outliers
```

```python
# Step 3: Feature Scaling
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
#scaler = StandardScaler()
```

```python
wcss=[]
for k in range(1,6):
    model=KMeans(n_clusters=k)
    model.fit(X)
    wcss.append(model.inertia_)
```

```python
# Plot elbow graph-
plt.plot(range(1,6),wcss)
plt.xlabel("No of cluster(k)")
plt.ylabel("WCSS")
plt.title("ElbowMethod")
```

```python
model=KMeans(n_clusters=2)
labels=model.fit_predict(X_scaled)
labels
```

```python
print("CLUSTERLABELS",labels)
```

```python
centroids = model.cluster_centers_#
print(centroids)
```

```python
score=silhouette_score(X_scaled, labels)
print("Silhouette Score:", score)
```

```python
plt.scatter(X,[0]*len(X),c=labels)
```
