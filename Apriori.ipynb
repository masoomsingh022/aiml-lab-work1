```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori,association_rules
```

```python
Transactions=[
    ['Bread','Butter','Milk'],
    ['Bread','Butter','Eggs'],
    ['Bread','Butter'],
     ['Bread','Milk'],
    ['Bread','Egg'],
    ['Egg','Milk'],
]
```

```python
df=pd.DataFrame({
    'Items':[' , '.join(i) for i in Transactions]
})
```

```python
df.to_csv("market_basket_dataset.csv",index=False)
pd.read_csv("market_basket_dataset.csv")
```

```python
dataset=[row.split(',') for row in df['Items']]
```

```python
model=TransactionEncoder()
te=model.fit_transform(dataset)
```

```python
new_df=pd.DataFrame(te,columns=model.columns_)
```

```python
frequent_items=apriori(
    new_df,
    min_support=0.3,
    use_colnames=True
)
print(frequent_items)
```

```python
rules=association_rules(
    frequent_items,
    metric='confidence',
    min_threshold=0.5
)
print(rules)
```

```python
print(rules[['antecedents','consequents','support','confidence']])
             
```
